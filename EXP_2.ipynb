```python
import socket

host = input("Enter target IP: ")
ports = [22, 80, 443, 8080]

print("Scanning", host)

for port in ports:
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.settimeout(1)

    try:
        result = s.connect_ex((host, port))

        if result == 0:
            print("Port", port, "is OPEN")
    except socket.gaierror:
        print("Invalid host")
        break
    finally:
        s.close()

print("Scan completed")
```
