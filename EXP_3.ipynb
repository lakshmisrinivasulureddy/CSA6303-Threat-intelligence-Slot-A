```python
policy = input("Enter security policy text: ").lower()

elements = {
    "Scope": "scope",
    "Roles and responsibilities": "roles",
    "Enforcement": "enforcement",
    "Review cycle": "review",
    "Approval authority": "approved by"
}

missing = []

for name, keyword in elements.items():
    if keyword not in policy:
        missing.append(name)

if missing:
    print("Missing elements:")
    for item in missing:
        print("-", item)
else:
    print("All required elements are present")
```
